```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Department', hue='Attrition')
plt.title('Attrition Count by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.show()
```

```python
attrition_rate = df.groupby('Department')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
print(attrition_rate.sort_values(ascending=False))
```

```python
plt.figure(figsize=(6,5))
sns.countplot(data=df, x='OverTime', hue='Attrition')
plt.title('Attrition Count by OverTime Status')
plt.xlabel('Works Overtime')
plt.ylabel('Number of Employees')
plt.show()
```

```python
overtime_rate = df.groupby('OverTime')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
print(overtime_rate)
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df, y='JobRole', hue='Attrition', order=df['JobRole'].value_counts().index)
plt.title('Attrition Count by Job Role')
plt.xlabel('Number of Employees')
plt.ylabel('Job Role')
plt.show()
```

```python
role_rate = df.groupby('JobRole')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
print(role_rate.sort_values(ascending=False))
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='Age', hue='Attrition', bins=20, kde=True, multiple='stack')
plt.title('Age Distribution by Attrition')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.show()
```

```python
df['AgeGroup'] = pd.cut(df['Age'], bins=[18, 25, 35, 45, 55, 60], labels=['18-25','26-35','36-45','46-55','56-60'])
age_rate = df.groupby('AgeGroup')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
print(age_rate)
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='MonthlyIncome', hue='Attrition', bins=25, kde=True, multiple='stack')
plt.title('Monthly Income Distribution by Attrition')
plt.xlabel('Monthly Income')
plt.ylabel('Number of Employees')
plt.show()
```

```python
df['IncomeQuartile'] = pd.qcut(df['MonthlyIncome'], 4, labels=['Q1 (Lowest)', 'Q2', 'Q3', 'Q4 (Highest)'])
income_rate = df.groupby('IncomeQuartile', observed=True)['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
print(income_rate)
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome')
plt.title('Monthly Income by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income')
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x='Attrition', y='Age')
plt.title('Age by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Age')
plt.show()
```

```python
plt.figure(figsize=(6,6))
df['Attrition'].value_counts().plot.pie(autopct='%1.1f%%', colors=['#4C72B0','#DD8452'], startangle=90)
plt.title('Overall Attrition Proportion')
plt.ylabel('')
plt.show()
```
